```python
# ============================================================
# CMD 06 — RECONSTRUCT CONSERVATIVE COMPOSITION VALIDITY
# ============================================================

# ------------------------------------------------------------
# Load elemental property table
# ------------------------------------------------------------
import re

atom_props_spark = spark.table(
    "ml_perovskites.atom_props"
)

atom_props_pd = atom_props_spark.toPandas()

print("atom_props shape:", atom_props_pd.shape)

# ------------------------------------------------------------
# Build valid element list
# ------------------------------------------------------------

VALID_ELEMENTS = set(
    atom_props_pd["Element"]
    .dropna()
    .astype(str)
    .str.strip()
)

print("Valid elements in atom_props:", len(VALID_ELEMENTS))

# ------------------------------------------------------------
# Conservative parser
# ------------------------------------------------------------

ELEMENT_REGEX = re.compile(
    r"([A-Z][a-z]?)([0-9]*\.?[0-9]*)"
)

VARIABLE_PATTERN = re.compile(
    r"[xyzXYZ]"
)

def parse_formula_conservative(formula):

    if pd.isna(formula):
        return {}, "missing"

    formula = str(formula).strip()

    if not formula:
        return {}, "missing"

    if VARIABLE_PATTERN.search(formula):
        return {}, "variable_stoichiometry"

    if "/" in formula:
        return {}, "special_notation"

    if any(
        token.lower() in formula.lower()
        for token in ["TBA", "SAC", "Gb"]
    ):
        return {}, "special_notation"

    matches = ELEMENT_REGEX.findall(formula)

    if not matches:
        return {}, "unparsed"

    composition = {}

    for element, amount in matches:

        if element not in VALID_ELEMENTS:
            return {}, "invalid_element"

        if amount == "":
            amount = 1.0
        else:
            amount = float(amount)

        composition[element] = (
            composition.get(element, 0.0)
            + amount
        )

    return composition, "valid"

# ------------------------------------------------------------
# Apply parser
# ------------------------------------------------------------

parsed_results = (
    db_ml[COMPOUND_COL]
    .apply(parse_formula_conservative)
)

db_ml["parsed_composition"] = (
    parsed_results
    .apply(lambda x: x[0])
)

db_ml["parse_status"] = (
    parsed_results
    .apply(lambda x: x[1])
)

db_ml["n_elements"] = (
    db_ml["parsed_composition"]
    .apply(len)
)

# ------------------------------------------------------------
# Audit
# ------------------------------------------------------------

print("\nParse status:")

display(
    db_ml["parse_status"]
    .value_counts()
    .to_frame("count")
)

print("\nTotal rows:", len(db_ml))

print(
    "Valid composition rows:",
    (db_ml["parse_status"] == "valid").sum()
)

print(
    "Invalid/non-modelable composition rows:",
    (db_ml["parse_status"] != "valid").sum()
)
```

```python
# ============================================================
# CMD 07 — GENERATE ELEMENTAL DESCRIPTORS
# ============================================================

# ------------------------------------------------------------
# Build elemental property lookup
# ------------------------------------------------------------

element_property_columns = [
    c for c in atom_props_pd.columns
    if c != "Element"
]

element_property_lookup = (
    atom_props_pd
    .set_index("Element")
    .to_dict(orient="index")
)

print(
    "Number of elemental properties:",
    len(element_property_columns)
)

# ------------------------------------------------------------
# Weighted descriptor function
# ------------------------------------------------------------

def weighted_descriptor(
    composition,
    property_name
):

    if not composition:
        return np.nan

    values = []
    weights = []

    for element, fraction in composition.items():

        if element not in element_property_lookup:
            continue

        raw_value = (
            element_property_lookup[element]
            .get(property_name)
        )

        value = pd.to_numeric(
            raw_value,
            errors="coerce"
        )

        if pd.isna(value):
            continue

        values.append(float(value))
        weights.append(float(fraction))

    if not values:
        return np.nan

    values = np.array(values)
    weights = np.array(weights)

    weight_sum = weights.sum()

    if weight_sum <= 0:
        return np.nan

    return np.sum(
        values * weights
    ) / weight_sum


# ------------------------------------------------------------
# Generate descriptors
# ------------------------------------------------------------

descriptor_data = {}

for property_name in element_property_columns:

    descriptor_name = f"mean_{property_name}"

    descriptor_data[descriptor_name] = (
        db_ml["parsed_composition"]
        .apply(
            lambda composition:
            weighted_descriptor(
                composition,
                property_name
            )
        )
    )

X_descriptors = pd.DataFrame(
    descriptor_data,
    index=db_ml.index
)

print("\nDescriptor matrix:")
print(X_descriptors.shape)

print(
    "\nExpected:",
    "(440, 58)"
)
```

```python
# ============================================================
# CMD 08 — APPLY VALIDATED FINAL DESCRIPTOR SET
# ============================================================

final_descriptor_candidates = [
    "mean_atomic_number",
    "mean_atomic_radius",
    "mean_atomic_radius_rahm",
    "mean_atomic_volume",
    "mean_boiling_point",
    "mean_bulk_modulus",
    "mean_covalent_radius_cordero",
    "mean_covalent_radius_pyykko_double",
    "mean_covalent_radius_pyykko_triple",
    "mean_density",
    "mean_dipole_polarizability",
    "mean_electron_negativity",
    "mean_electron_affinity",
    "mean_en_ghosh",
    "mean_first_ion_en",
    "mean_fusion_enthalpy",
    "mean_gs_bandgap",
    "mean_gs_energy",
    "mean_gs_est_bcc_latcnt",
    "mean_gs_mag_moment",
    "mean_hhi_p",
    "mean_hhi_r",
    "mean_heat_capacity_mass",
    "mean_heat_capacity_molar",
    "mean_icsd_volume",
    "mean_heat_of_formation",
    "mean_lattice_constant",
    "mean_mendeleev_number",
    "mean_melting_point",
    "mean_num_unfilled",
    "mean_num_valance",
    "mean_num_d_unfilled",
    "mean_num_d_valence",
    "mean_num_f_valence",
    "mean_num_p_unfilled",
    "mean_num_p_valence",
    "mean_num_s_unfilled",
    "mean_period",
    "mean_specific_heat",
    "mean_thermal_conductivity",
    "mean_vdw_radius_mm3",
    "mean_vdw_radius_uff",
    "mean_sound_velocity"
]

# ------------------------------------------------------------
# Validate feature availability
# ------------------------------------------------------------

missing_features = [
    c for c in final_descriptor_candidates
    if c not in X_descriptors.columns
]

if missing_features:
    raise RuntimeError(
        f"Final descriptors missing from generated matrix: "
        f"{missing_features}"
    )

# ------------------------------------------------------------
# Create reduced descriptor matrix
# ------------------------------------------------------------

X_reduced = X_descriptors[
    final_descriptor_candidates
].copy()

# ------------------------------------------------------------
# Validation
# ------------------------------------------------------------

print("Final descriptor matrix:")
print(X_reduced.shape)

print("\nNumber of descriptors:")
print(X_reduced.shape[1])

print("\nNumber of rows:")
print(X_reduced.shape[0])

print("\nUnique descriptors:")
print(len(set(final_descriptor_candidates)))

assert X_reduced.shape == (440, 43)
assert len(set(final_descriptor_candidates)) == 43

print("\nValidation passed.")
```

```python
#===============
# TEMPORARY
#===============

# ============================================================
# DIAGNOSTIC — H₂ TARGET DISTRIBUTION BY COMPOUND
# ============================================================

h2_compound_summary = (
    hydrogen_data
    .groupby(COMPOUND_COL)["hydrogen_target"]
    .agg(
        n_records="count",
        min_h2="min",
        median_h2="median",
        max_h2="max"
    )
    .reset_index()
    .sort_values("max_h2", ascending=False)
)

print("=" * 70)
print("TOP COMPOUNDS BY MAXIMUM H₂")
print("=" * 70)

display(
    h2_compound_summary.head(20)
)

print("\nNumber of compounds with H₂ >= 10,000:")
print(
    (h2_compound_summary["max_h2"] >= 10000).sum()
)

print("\nNumber of compounds with H₂ >= 100,000:")
print(
    (h2_compound_summary["max_h2"] >= 100000).sum()
)

print("\nNumber of compounds with H₂ >= 500,000:")
print(
    (h2_compound_summary["max_h2"] >= 500000).sum()
)
```

```python
# ======================================================================
# CMD 16 — COMPOUND-GROUPED + TARGET-STRATIFIED H₂ SPLIT
# ======================================================================

from sklearn.model_selection import train_test_split
import numpy as np
import pandas as pd

# ----------------------------------------------------------------------
# Build one row per compound
# ----------------------------------------------------------------------

h2_group_summary = (
    hydrogen_data
    .groupby(COMPOUND_COL)["hydrogen_target"]
    .agg(
        n_records="count",
        max_h2="max"
    )
    .reset_index()
)

# ----------------------------------------------------------------------
# Define target strata at COMPOUND level
# ----------------------------------------------------------------------

def assign_h2_stratum(x):

    if x == 0:
        return "zero"
    elif x < 100:
        return "very_low"
    elif x < 1000:
        return "low"
    elif x < 10000:
        return "medium"
    elif x < 100000:
        return "high"
    else:
        return "extreme"


h2_group_summary["target_stratum"] = (
    h2_group_summary["max_h2"]
    .apply(assign_h2_stratum)
)

print("=" * 70)
print("H₂ COMPOUND-LEVEL TARGET STRATA")
print("=" * 70)

display(
    h2_group_summary["target_stratum"]
    .value_counts()
    .sort_index()
)

# ----------------------------------------------------------------------
# Split COMPOUNDS, not individual observations
# ----------------------------------------------------------------------

train_compounds, test_compounds = train_test_split(
    h2_group_summary,
    test_size=0.20,
    random_state=42,
    stratify=h2_group_summary["target_stratum"]
)

train_compound_names = set(
    train_compounds[COMPOUND_COL]
)

test_compound_names = set(
    test_compounds[COMPOUND_COL]
)

# ----------------------------------------------------------------------
# Apply compound split to all H₂ observations
# ----------------------------------------------------------------------

hydrogen_train = hydrogen_data[
    hydrogen_data[COMPOUND_COL].isin(train_compound_names)
].copy()

hydrogen_test = hydrogen_data[
    hydrogen_data[COMPOUND_COL].isin(test_compound_names)
].copy()

# ----------------------------------------------------------------------
# Validation
# ----------------------------------------------------------------------

print("\n" + "=" * 70)
print("HYDROGEN — GROUPED + TARGET-STRATIFIED SPLIT")
print("=" * 70)

print("Train rows:", len(hydrogen_train))
print("Test rows:", len(hydrogen_test))

print(
    "Train unique compounds:",
    hydrogen_train[COMPOUND_COL].nunique()
)

print(
    "Test unique compounds:",
    hydrogen_test[COMPOUND_COL].nunique()
)

print(
    "Train percentage:",
    round(len(hydrogen_train) / len(hydrogen_data) * 100, 2)
)

print(
    "Test percentage:",
    round(len(hydrogen_test) / len(hydrogen_data) * 100, 2)
)

print("\nMaximum H₂:")
print("Train:", hydrogen_train["hydrogen_target"].max())
print("Test :", hydrogen_test["hydrogen_target"].max())

print("\nHigh-value compounds:")
print(
    "Train compounds >= 100,000:",
    (train_compounds["max_h2"] >= 100000).sum()
)

print(
    "Test compounds >= 100,000:",
    (test_compounds["max_h2"] >= 100000).sum()
)
```

```python
# ======================================================================
# CMD 17 — VERIFY ZERO COMPOUND OVERLAP
# ======================================================================

print("=" * 70)
print("HYDROGEN — GROUP LEAKAGE CHECK")
print("=" * 70)

train_compounds_h2 = set(
    hydrogen_train[COMPOUND_COL].unique()
)

test_compounds_h2 = set(
    hydrogen_test[COMPOUND_COL].unique()
)

overlap_h2 = train_compounds_h2.intersection(
    test_compounds_h2
)

print("Train compounds:", len(train_compounds_h2))
print("Test compounds:", len(test_compounds_h2))
print("Overlapping compounds:", len(overlap_h2))

if len(overlap_h2) == 0:
    print("PASS — No compound appears in both train and test.")
else:
    print("FAIL — Compound leakage detected:")
    print(sorted(overlap_h2))
```

```python
# ======================================================================
# CMD 18 — FINAL H₂ MODELING MATRICES
# ======================================================================

print("=" * 70)
print("FINAL H₂ MODELING MATRICES")
print("=" * 70)

# H₂ target
y_train_h2 = hydrogen_train["hydrogen_target"].copy()
y_test_h2  = hydrogen_test["hydrogen_target"].copy()

# Feature columns
exclude_cols = {
    "hydrogen_target",
    "bandgap_target",
    COMPOUND_COL
}

h2_feature_cols = [
    c for c in hydrogen_train.columns
    if c not in exclude_cols
]

X_train_h2 = hydrogen_train[h2_feature_cols].copy()
X_test_h2  = hydrogen_test[h2_feature_cols].copy()

print("H₂ X_train:", X_train_h2.shape)
print("H₂ X_test :", X_test_h2.shape)
print("H₂ y_train:", y_train_h2.shape)
print("H₂ y_test :", y_test_h2.shape)

# ----------------------------------------------------------------------
# Missing-value audit
# ----------------------------------------------------------------------

train_missing = X_train_h2.isna().sum()
train_missing = train_missing[train_missing > 0]

test_missing = X_test_h2.isna().sum()
test_missing = test_missing[test_missing > 0]

print("\nMissing values — X_train:")
print(train_missing if len(train_missing) else "0")

print("\nMissing values — X_test:")
print(test_missing if len(test_missing) else "0")

# ----------------------------------------------------------------------
# Show affected training rows
# ----------------------------------------------------------------------

if len(train_missing) > 0:

    missing_features = train_missing.index.tolist()

    print("\nAffected training rows:")
    display(
        hydrogen_train.loc[
            X_train_h2[missing_features].isna().any(axis=1),
            [COMPOUND_COL, "hydrogen_target"] + missing_features
        ]
    )
```
