```python
# hydrogen_dataset_builder.py

import pandas as pd

INPUT_FILE = "filtered_feature_matrix.csv"
OUTPUT_FILE = "hydrogen_feature_matrix.csv"

# load
df = pd.read_csv(INPUT_FILE)

# ----------------------------------
# REMOVE RAW H2 TARGET
# ----------------------------------

if "hydrogen_production_rate" in df.columns:
    df = df.drop(columns=["hydrogen_production_rate"])

# ----------------------------------
# KEEP ONLY ROWS WITH H2 TARGET
# ----------------------------------

target = "log_hydrogen_production_rate"

df_h2 = df[
    df[target].notna()
].copy()

# ----------------------------------
# SAVE
# ----------------------------------

df_h2.to_csv(
    OUTPUT_FILE,
    index=False
)

print("Hydrogen dataset created.")
print("Shape:", df_h2.shape)
print("Output:", OUTPUT_FILE)

print("\nColumns:")
print(df_h2.columns.tolist())

print("\nTarget summary:")
print(df_h2[target].describe())

missing = df_h2.isna().sum()
print("\nColumns with missing values:")
print(missing[missing > 0].sort_values(ascending=False))
```
