```python
# bandgap_dataset_builder.py

import pandas as pd

INPUT_FILE = "filtered_feature_matrix.csv"
OUTPUT_FILE = "bandgap_feature_matrix.csv"

# load
df = pd.read_csv(INPUT_FILE)

# ----------------------------------
# TARGET
# ----------------------------------

target = "bandgap_energy"

# ----------------------------------
# REMOVE EXPERIMENTAL DESCRIPTORS
# ----------------------------------

experimental_cols = [
    "radiation_source_intensity",
    "calcination_temperature", 
    "catalyst_amount"
]

# keep only existing columns
experimental_cols = [
    c for c in experimental_cols
    if c in df.columns
]

# remove experimental variables
df_bandgap = df.drop(
    columns=experimental_cols,
    errors="ignore"
)

# ----------------------------------
# REMOVE H2 TARGET
# ----------------------------------

for col in [
    "hydrogen_production_rate",
    "log_hydrogen_production_rate"
]:
    if col in df_bandgap.columns:
        df_bandgap = df_bandgap.drop(columns=col)

# ----------------------------------
# KEEP ONLY ROWS WITH EXPERIMENTAL Eg
# ----------------------------------

df_bandgap = df_bandgap[
    df_bandgap[target].notna()
]

# ----------------------------------
# SAVE
# ----------------------------------

df_bandgap.to_csv(
    OUTPUT_FILE,
    index=False
)

print("Bandgap dataset created.")
print("Shape:", df_bandgap.shape)
print("Output:", OUTPUT_FILE)

print("\nColumns:")
print(df_bandgap.columns.tolist())
```
